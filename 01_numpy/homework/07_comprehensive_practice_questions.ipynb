````python
import numpy as np

# 5개 지점의 월별 사과, 바나나, 오렌지 판매량 (5x3x6 배열)
# 5개의 지점, 3개의 과일(사과, 바나나, 오렌지), 6개월치 데이터
sales_data = np.array([
    # 지점 1
    [[120, 150, 200, 230, 180, 210],  # 사과 판매량
     [100, 140, 170, 190, 160, 180],  # 바나나 판매량
     [90, 120, 160, 140, 130, 150]],  # 오렌지 판매량

    # 지점 2
    [[80, 90, 110, 140, 130, 120],
     [60, 70, 90, 100, 110, 95],
     [50, 80, 100, 110, 90, 85]],

    # 지점 3
    [[130, 140, 160, 180, 170, 190],
     [110, 120, 130, 150, 140, 160],
     [100, 110, 120, 130, 120, 140]],

    # 지점 4
    [[100, 130, 150, 170, 160, 180],
     [90, 100, 110, 120, 130, 125],
     [80, 110, 140, 130, 120, 110]],

    # 지점 5
    [[90, 110, 130, 140, 120, 130],
     [70, 85, 100, 110, 105, 95],
     [60, 90, 110, 100, 90, 95]]
])

fruits = np.array(["사과", "바나나", "오렌지"])
branches = np.array(["지점1", "지점2", "지점3", "지점4", "지점5"])
months = np.array(["1월", "2월", "3월", "4월", "5월", "6월"])
````

````python
# 1. 각 지점에서 6개월 동안 판매된 **사과**, **바나나**, **오렌지**의 월별 평균 판매량을 계산하세요.
branch_avgs = np.mean(sales_data, axis=2)

for b_index, b_name in enumerate(branches):
    print(b_name)
    for f_index, f_name in enumerate(fruits):
        print(f'{f_name} : {branch_avgs[b_index][f_index]:.2f}')
````

````python
# 2. 각 월마다 전체 지점에서 가장 많이 판매된 과일을 찾아 판매량과 과일명을 출력하세요.
sales_monthly = np.sum(sales_data, axis=0)
# print(sales_monthly)

sales_monthly_most = np.argmax(sales_monthly, axis=0)
# print(sales_monthly_most)

for idx, month in enumerate(months):
    f_idx = sales_monthly_most[idx]
    f_name = fruits[f_idx]
    print(f'{month} - {f_name} : {sales_monthly[f_idx, idx]}')
````

````python
# 3. 6개월 동안 전체 지점에서 가장 많이 판매된 과일의 총량을 과일별로 구하세요. 어떤 과일이 가장 많이 판매되었는지 확인하세요.

sales_most_total = np.sum(sales_monthly, axis=1)
# print(sales_most_total)

sales_the_most = np.argmax(sales_most_total, axis=0)

print(f'{fruits[sales_the_most]} : {sales_most_total[sales_the_most]}')
````

````python
# 4. 각 지점에서 판매량이 150개 이상인 데이터를 추출하고 그에 해당하는 지점과 과일을 출력하세요.

check_cond = sales_data >= 150
# print(check_cond)

for b_idx, b_name in enumerate(branches):
    for f_idx, f_name in enumerate(fruits):
        for m_idx, month in enumerate(months):
            if check_cond[b_idx, f_idx, m_idx]:
                print(f'{b_name} - {month} : {f_name}')
````
